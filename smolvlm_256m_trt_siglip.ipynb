# SmolVLM-256M: TensorRT Engine for SigLIP

This notebook tests a TensorRT engine for the SigLIP vision encoder. The prefill profile showed that SigLIP uses 68% of the prefill.

The script does these steps:

1. It runs the model and saves the input of the SigLIP encoder.
2. It exports the encoder (the 12 transformer layers and the last layer norm) to ONNX.
3. It builds an FP16 TensorRT engine. It uses `trtexec` if the tool is on the path. If not, it uses the TensorRT Python API.
4. It replaces the SigLIP module of the model with the engine. The patch embedding stays in PyTorch.

The engine has a fixed shape for the tile count of the test image (13 tiles). An engine is tied to the GPU and to the TensorRT version. Build it on the machine that runs it. For the Jetson, build it on the Orin.

The script prints the build times. Then it compares PyTorch and TensorRT: the prefill time, the SigLIP time, the speed-up, the difference in the encoder output, and whether the answer is the same.
