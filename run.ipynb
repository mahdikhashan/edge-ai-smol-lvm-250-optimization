# SmolVLM-256M: CUDA Graph Decoding

This notebook shows a method to make SmolVLM-256M generate text faster on a GPU.

The model works in two phases. In the prefill phase, the model reads the image and the prompt. In the decode phase, the model generates one token at a time.

Each decode step starts many small GPU kernels. In eager mode, the CPU can spend more time on these starts than the GPU spends on the work.

The method records one decode step as a CUDA graph. The GPU then replays the graph for each token. One replay is one CPU call.

The graph uses a static KV cache and fixed tensors, so each step has the same shape. The prefill phase stays eager, because its shape changes with the number of image tiles.

The notebook shows the image and the prompt. It then runs the script on the image. Last, it compares eager decoding and graph decoding, and checks that both give the same tokens.
