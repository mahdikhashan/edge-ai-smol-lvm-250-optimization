# SmolVLM-256M: Fewer Image Tiles

This notebook tests how the number of image tiles changes the speed and the answer. The profile showed that each tile costs about 17 ms of prefill, and that the prefill is 74% of the request.

The script runs a range of settings on the same image:

- `default` uses the normal image splitting.
- `sizeN` resizes the longest side of the image to N pixels before the image is cut into 512-pixel tiles. A smaller N gives fewer tiles.
- `nosplit` turns the image splitting off.

For each setting, the script prints the number of tiles, the number of prompt tokens, and the cache length. It uses the smallest cache that fits the prompt. It also prints the time of the input, the prefill and the decode (with stop at EOS), the total time of the request, the speed-up against `default`, and the answer. It shows if the answer is the same as the answer of `default`.
